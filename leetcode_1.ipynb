```python
import os
from getpass import getpass

# 1. Manual Prompt via getpass
print("🔑 Please enter your LeetCode credentials:")
user_name = getpass("LEETCODE_USERNAME: ")
user_cookie = getpass("LEETCODE_TOKEN: ")

# 2. Setup - We run these via '!' but avoid %%capture at the very top 
# so the getpass prompt can actually be seen by the user.
!npm install -g leetcode-tools/leetcode-cli > /dev/null 2>&1
!git clone https://github.com/aqwertyuiop48/leetcode_programming.git || true


if user_name and user_cookie:
    # 3. Assign to OS environment
    os.environ['LEETCODE_USERNAME'] = user_name
    os.environ['LEETCODE_TOKEN'] = user_cookie

    print("✅ Credentials received. Attempting login...")

    # 4. Run the login
    # We use -q or redirect output to keep it clean if desired
    !export PATH=$PATH:$(npm config get prefix)/bin && \
     printf "%s\n%s\n" "$LEETCODE_USERNAME" "$LEETCODE_TOKEN" | leetcode user -c
else:
    print("❌ Error: Username or Token was left empty.")
```
